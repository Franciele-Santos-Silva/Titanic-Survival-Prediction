A base teste possui colunas com valores vazios que não estão na base de treino, nesse caso precisariamos tratar essas colunas apenas na base de teste

Colunas com alta cardinalidade não vão ajudar o modelo pois podem torna-ló menos generalizadável, nesse caso, a melhor alternativa é remover essas colunas do conjunto de dados.

```python
#eliminando para a base de treino
treino = treino.drop(['Name','Ticket','Cabin'], axis=1)
```

```python
#eliminando para a base de teste
teste = teste.drop(['Name','Ticket','Cabin'], axis=1)
```

Agora a coluna Age será tratada, pois possui valores vazios tanto em de treino quanto em teste. Como essa coluna apresenta a idade dos passageiros, utilizarei a média das idades para substituir os valores vazios.

```python
# verificar a média das idades da base de treino
treino.Age.mean()
```

```python
#selecionar as colunas com idade vazia para substituir por essa média na base de treino
treino.loc[treino.Age.isnull(), 'Age'] = treino.Age.mean()
```

```python
# #selecionar as colunas com idade vazia para substituir por essa média na base de teste
teste.loc[teste.Age.isnull(), 'Age'] = teste.Age.mean()
```

```python
#verificar os valores na coluna Embarked
treino.Embarked .value_counts()
```

```python
#moda da coluna
treino.Embarked.mode()[0]
```

```python
#substituir valores vazios
treino.loc[treino.Embarked.isnull(), 'Embarked'] = treino.Embarked.mode()[0]
```

```python
#varificar a média da coluna Fare para a base de teste
teste.loc[teste.Fare.isnull(),'Fare']= teste.Fare.mean()
```
